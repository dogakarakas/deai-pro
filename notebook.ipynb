# Ideas for Pipelines

<b>Domain feature additions</b>
- Add engineered features built from knowledge of the dataset, then see if the score improves over the baseline. Examples already computed in your pipeline_1_1 template: TotalSpend, Group_Size, Is_Child, Cabin_Deck, Family_Size. “Domain” means the ideas come from understanding the problem (people in the same group travel together, children don’t spend money, and so on), not from the model.


<b>Feature ablations</b>
- An ablation means removing something to measure how much it mattered. Start with a feature set, drop one feature or one group of features at a time, and re-score. If accuracy falls a lot, that feature was important. If it doesn’t change, it’s redundant. This is the natural follow-up to adding features: it tells you which additions actually helped.

<b>CPU random forests</b>
- A random forest trains many decision trees, each on a random sample of rows and features, and averages their votes. It handles nonlinear patterns and interactions between features (for example, “CryoSleep passengers never spend money”) that logistic regression can’t capture. “CPU” means it runs on the processor, not a GPU, so it works on your laptop.

<b>CPU histogram boosting</b>

- Gradient boosting builds trees one after another, where each new tree corrects the mistakes of the previous ones. “Histogram” refers to a speed trick: continuous features are bucketed into bins, which makes training much faster. This is the family your pyproject.toml already installs: LightGBM, XGBoost, CatBoost (and scikit-learn’s HistGradientBoostingClassifier). On tabular data like this, boosting is usually the strongest single model type.

<b>Fixed probability blends</b>
- Blending combines the predicted probabilities of several models, for example averaging them, because different models make different mistakes. “Fixed” means the blend weights are chosen in advance (like a plain 50/50 average) rather than learned by another model. This is where saving oof_predictions.csv pays off: the blend takes the probability files from earlier pipelines and averages them, then scores the result against the true labels.
